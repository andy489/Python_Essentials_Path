# Mapping with Dictionaries

### Creating Dictionaries

```python
settings_dict = {
    "resolution":"1920x1080", 
    "fullscreen": True,
    "volume": 75,
}

print("Settings dictionary:", settings_dict)
print("Length of a dictionary:", len(settings_dict))

empty_dict = {}
print("Empty dictionary: ", empty_dict)
```

### Initializing Dictionary Values with `fromkeys()`

```python
permissions = ["read", "write", "delete", "export"]
default_permissions = dict.fromkeys(permissions, False)

print(default_permissions)
```

### Creating a Dictionary from Two Lists with `zip()`

```python
keys   = ["host",       "port", "protocol"]
values = ("db.internal", 5432,  "tcp")

# zip() pairs elements by position, producing one tuple per index
for pair in zip(keys, values):
    print(pair)

# Equivalent to:
# ("host",     "db.internal")
# ("port",      5432)
# ("protocol", "tcp")
```

```python
keys = ["host", "port", "protocol"]
values = ("db.internal", 5432, "tcp")   # tuple works just as well as a list

# zip() pairs each key with its corresponding value
connection = dict(zip(keys, values))
print("Connection config:", connection)

# If the sequences have different lengths, zip() stops at the shorter one
extra_keys = ["host", "port", "protocol", "timeout"]
short_values = (5432, "tcp")
truncated = dict(zip(extra_keys, short_values))
print("Truncated (2 pairs):", truncated)
```

### Accessing Dictionary Items with Keys

```python
config = {
    "resolution": "1920x1080",
    "fullscreen": True,
}

print(config["resolution"])
```

### Nested Dictionaries

```python
user_preferences = {
    "userX": {
        "email": "userx@example.com",
        "preferences": {
            "theme": "dark",
            "notifications": True
        }
    },
    "userY": {
        "email": "usery@example.com",
        "preferences": {
            "theme": "light",
            "notifications": False
        }
    }
}

print("User X theme:", user_preferences["userX"]["preferences"]["theme"])
```

### Accessing Non-existent Keys will Raise a `KeyError`

```python
config = {
    "resolution": "1920x1080",
    "fullscreen": True,
}

try:
    print(config["brightness"]) # This key doesn't exist
except KeyError:
    print("The 'brightness' setting is missing.")
```

### Checking if the Key Exists

```python
config = {
    "resolution": "1920x1080",
    "fullscreen": True,
}

# Check if the key is NOT present
if "brightness" not in config:
    print("The 'brightness' setting is missing.")
else:
    print("Brightness is:", config["brightness"])

# Check if the key is present
if "resolution" in config:
    print("Resolution is:", config["resolution"])
else:
    print("The 'resolution' setting is missing.")
```

### Accessing Elements with `get()`

```python
config = {
    "resolution": "1920x1080",
    "fullscreen": True,
}

resolution = config.get("resolution", "[MISSING: Set default resolution]")
print("Resolution:", resolution)

brightness = config.get("brightness", "[MISSING: Set default brightness]") # Default to a warning message
print("Brightness:", brightness)

language = config.get("language", "en") # Default to English
print("Selected language:", language)

print("Theme:", config.get("theme")) # This non-existent key will NOT raise a KeyError
```

### Applying Default Values with `setdefault()`

```python
user = {
    "username": "data_builder",
}

user_role = user.setdefault("role", "viewer") # If the key doesn't exist, add it with the default value
print("User's role:", user_role)

username = user.setdefault("username", "data_tester") # If the key exists, return its corresponding value
print("Username:", username)

print(user)
```

### Adding and Updating Items

```python
sensors = {
    "temperature": "22°C"
}

sensors["humidity"] = "60%" # Add new sensor reading
sensors["temperature"] = "23°C" # Update reading

print(sensors)
```

### Updating a Dictionary with `update()`

```python
global_settings = {
    "sampling_rate": 60,
    "units": "metric",
    "precision": 2,
}

device_overrides = {
    "precision": 3,             # existing key
    "units": "imperial",        # existing key
    "calibration_offset": 0.05  # new key
}

global_settings.update(device_overrides)

print(global_settings)
```

### Merging Dictionaries with the `|` (Merge) Operator

```python
global_settings = {
    "sampling_rate": 60,
    "units": "metric",
    "precision": 2,
}

device_overrides = {
    "precision": 3,             # existing key
    "units": "imperial",        # existing key
    "calibration_offset": 0.05  # new key
}

# Returns a new dictionary
new_settings = global_settings | device_overrides

print(new_settings)
```

### Updating a Dictionary with the `|=` (Merge-update) Operator

```python
global_settings = {
    "sampling_rate": 60,
    "units": "metric",
    "precision": 2,
}

device_overrides = {
    "precision": 3,             # existing key
    "units": "imperial",        # existing key
    "calibration_offset": 0.05  # new key
}

global_settings |= device_overrides

print(global_settings)
```

### Merging Dictionaries Inside a Dict Literal with `**`

```python
defaults = {"theme": "light", "language": "en", "timeout": 30}
overrides = {"theme": "dark", "timeout": 60}

# {**a, **b} spreads both dicts into a new one — rightmost value wins on collision
# Works anywhere a dict literal is valid (function calls, assignments, comprehensions)
merged = {**defaults, **overrides}
print("Merged:", merged)

# Handy for adding extra keys inline without mutating the originals
with_debug = {**merged, "debug": True}
print("With debug:", with_debug)
print("Originals unchanged:", defaults["theme"], overrides["theme"])
```

### Removing Items with `del`

```python
sensors = {
    "temperature": "22°C",
    "humidity": "60%",
    "pressure": "1013 hPa",
}

# If the key doesn't exist, del will raise a KeyError
del sensors["humidity"]
    
print(sensors)
```

### Removing Items with `pop()`

```python
sensors = {
    "temperature": "22°C",
    "humidity": "60%",
    "pressure": "1013 hPa"
}

# pop() will return the value of the removed element
value = sensors.pop("pressure")

print(value)
print(sensors)
```

### Safe Removal with `pop()` Default Value

```python
sensors = {
    "temperature": "22°C",
    "humidity": "60%",
}

# Without a default, pop() raises KeyError if the key is missing
try:
    sensors.pop("pressure")
except KeyError as e:
    print("KeyError:", e)

# With a default, pop() returns the default instead of raising
removed = sensors.pop("pressure", None)
print("Removed:", removed)      # None — key wasn't there, no error

# Useful when cleaning up optional keys
timeout = sensors.pop("timeout", 30)
print("Timeout setting:", timeout)  # falls back to 30
```

### Removing Items with `popitem()`

```python
sensors = {
    "temperature": "22°C",
    "humidity": "60%",
    "pressure": "1013 hPa"
}

# popitem() will remove and return last key-value pair
last_element = sensors.popitem()

print(last_element)
print(sensors)
```

### Clearing the Dictionary with `clear()`

```python
sensors = {
    "temperature": "22°C",
    "humidity": "60%",
    "pressure": "1013 hPa"
}

sensors.clear()
print(sensors)
```

### Creating Dictionary View Objects with `items()`

```python
prices = {
    "keyboard": 29.99,
    "monitor": 189.99,
    "mouse": 19.99,
    "chair": 120.00,
}

print(prices.items())

alphabetical = dict(sorted(prices.items()))
print(alphabetical)

# item is each (key, value) tuple from items(); item[1] selects the value
sorted_prices = dict(sorted(prices.items(), key=lambda item: item[1]))
print(sorted_prices)
```

### Sorting Dictionaries by Value with `itemgetter()`

```python
from operator import itemgetter

prices = {
    "keyboard": 29.99,
    "monitor": 189.99,
    "mouse": 19.99,
    "chair": 120.00,
}

sorted_prices_getter = dict(sorted(prices.items(), key=itemgetter(1))) # itemgetter is a compiled C function (faster than python interpreted)
print("Sorted with itemgetter:", sorted_prices_getter)
```

### Creating Dictionary View Objects with `values()`

```python
uptime_hours = {
    "server1": 120,
    "server2": 98,
    "server3": 143,
    "server4": 0, # offline
}

print(uptime_hours.values())
```

### Aggregate Functions

```python
values = uptime_hours.values()

print("Total servers:", len(values))       # Total number of servers
print("Total uptime:", sum(values))        # Combined uptime
print("Max uptime:", max(values))          # Longest running server
print("Min uptime:", min(values))          # Shortest (could be 0)
print("All running?", all(values))         # False if any value is 0 or falsy
print("Any running?", any(values))         # True if at least one is > 0
```

### Transforming Dictionaries with Dictionary Comprehensions

```python
# File sizes in MB
file_sizes = {
    "report.pdf": 4,
    "photo.png": 2,
    "data.csv": 12,
}

# Convert file sizes to KB
sizes_kb = {filename: size * 1024 for filename, size in file_sizes.items()}

print(sizes_kb)
```

### Filtering Dictionaries with Dictionary Comprehensions

```python
inventory = {
    "apple":  {"qty": 50,  "price": 0.5},
    "banana": {"qty": 0,   "price": 0.3},
    "cherry": {"qty": 120, "price": 1.2},
    "date":   {"qty": 0,   "price": 2.5},
}

# Keep only items that are in stock
in_stock = {item: info for item, info in inventory.items() if info["qty"] > 0}
print("In stock:", list(in_stock.keys()))

# Keep only items that are in stock AND cost under $1
affordable = {
    item: info
    for item, info in inventory.items()
    if info["qty"] > 0 and info["price"] < 1.0
}
print("Affordable in-stock:", list(affordable.keys()))
```

### Making a Shallow Copy of a Dictionary with `copy()`

```python
original = {"theme": "dark"}

shallow_copy = original.copy()
# You can also use dict()
# shallow_copy = dict(original)

shallow_copy["theme"] = "light"

print("original:", original)
print("shallow_copy:", shallow_copy)
```

### Making a Deep Copy of a Dictionary with `deepcopy()`

```python
import copy

original = {
    "theme": "dark",
    "options": {"autosave": True},
}

deep_copy = copy.deepcopy(original)

deep_copy["options"]["autosave"] = False

print("original:", original)
print("deep_copy:", deep_copy)
```

### Unpacking Dictionaries with the ** Operator

```python
def greet(greeting, name):
    print(f"{greeting}, {name}!")

# Positional arguments
greet("Welcome", "Guest")

# Keyword Arguments
greet(greeting="Good evening", name="Visitor")

kwargs = {
    "greeting": "Hello",
    "name": "Traveller",
}

greet(**kwargs)
# This is equal to calling:
# greet(greeting=kwargs["greeting"], name=kwargs["name"])
```

### Passing Wrong Parameter Names Will Result in a `TypeError`

```python
def greet(greeting, name):
    print(f"{greeting}, {name}!")

wrong_kwargs = {"salutation": "Hey", "name": "Guest"} # 'salutation' doesn't match parameter name

try:
    greet(**wrong_kwargs)
except TypeError as e:
    print("TypeError caught:", e)
```

### Packing Keyword Arguments with the `**` Operator

```python
def log_event(event_type, **metadata):
    print("Event:", event_type)
    print("Details:", metadata)

log_event("login", user="anon", location="unknown")
```

### Example of Using Keyword Arguments

```python
def html_tag(tag, content, **kwargs):
    """
    Returns a string representation of an HTML tag.

    :param tag: The HTML tag name.
    :param content: The content inside the tag.
    :param kwargs: Optional HTML attributes for the tag.
    """
    print("kwargs:", kwargs)
    attributes = ' '.join([f'{key}="{value}"' for key, value in kwargs.items()])
    return f'<{tag} {attributes}>{content}</{tag}>'

# Example usage
print(html_tag('a', 'Click here', href="https://example.com", style="color: red;"))
```

## Advanced Topics

### `defaultdict` — Automatic Default Values

`collections.defaultdict` works like a regular dict but automatically inserts a default value when a missing key is accessed, eliminating the need for explicit `setdefault()` or `if key not in d` guards.

```python
from collections import defaultdict

# Group server names by data centre — with a plain dict you'd need:
# if dc not in groups: groups[dc] = []
# groups[dc].append(server)
servers = [
    ("eu-west", "web-01"), ("eu-west", "db-01"),
    ("us-east", "web-02"), ("us-east", "cache-01"), ("us-east", "db-02"),
]

groups = defaultdict(list)   # missing key → empty list automatically
for dc, server in servers:
    groups[dc].append(server)

for dc, names in groups.items():
    print(f"{dc}: {names}")

# Other common factory types
counts  = defaultdict(int)   # missing key → 0
flags   = defaultdict(bool)  # missing key → False
nested  = defaultdict(dict)  # missing key → {}

counts["errors"] += 1
print("\nError count:", counts["errors"])
print("Missing key:", counts["never_set"])  # 0, not KeyError
```

### `Counter` — Frequency Counting

`collections.Counter` is a dict subclass designed for counting hashable objects. It accepts any iterable and provides helpers like `most_common()` and arithmetic between counters.

```python
from collections import Counter

# Count log-level occurrences from a stream of log entries
log_levels = ["INFO", "DEBUG", "ERROR", "INFO", "WARNING", "ERROR", "INFO", "ERROR"]

level_counts = Counter(log_levels)
print("All counts:", dict(level_counts))
print("Top 2 levels:", level_counts.most_common(2))

# Counter works on strings (character frequency) and anything iterable
word = "mississippi"
char_freq = Counter(word)
print("\nChar frequencies:", char_freq.most_common())

# Arithmetic: combine counts from two time windows
morning = Counter({"INFO": 40, "ERROR": 5, "WARNING": 3})
evening = Counter({"INFO": 25, "ERROR": 12, "WARNING": 1})

combined = morning + evening
print("\nCombined:", dict(combined))

# Subtract counts
delta = evening - morning   # only keeps positive results
print("New errors in evening vs morning:", dict(delta))
```
