# Working with Tuples
Ordered immutable collection of elements

### Creating Tuples

```python
empty_tuple = ()
print(empty_tuple)

# Trailing comma is required
one_element_tuple = (42,)
print(one_element_tuple)

multiple_elements = (1, "apple", 3.14)
print(multiple_elements)
print("Length of a tuple:", len(multiple_elements))

# Tuple without parentheses
implicit_tuple = 10, 20, 30
print(implicit_tuple)
```

### Accessing Tuple Elements

```python
# RGB color for a soft purple
rgb_color = (150, 100, 200)

print("RGB Color:", rgb_color)
print("Red component:", rgb_color[0])
print("Green component:", rgb_color[1])
print("Blue component:", rgb_color[2])
```

### Tuples are Immutable

```python
rgb_color = (150, 100, 200)

# Attempt to change the second element
try:
    rgb_color[1] = 10
except TypeError as e:
    print("Error:", e)
```

```python
def calculate_distance(coord1, coord2):
    # A simple Euclidean distance calculation for demonstration
    return ((coord1[0] - coord2[0]) ** 2 + (coord1[1] - coord2[1]) ** 2) ** 0.5

# Coordinates as tuples
point_a = (40.7128, -74.0060)  # New York
point_b = (34.0522, -118.2437) # Los Angeles

distance = calculate_distance(point_a, point_b)
print(f"Distance: {distance} units")
```

### Tuple Operations

```python
# Temperature readings from different regions
north_region = (21.5, 23.0)
south_region = (19.8, 22.1)
east_region = (21.5, 24.0)

# Concatenation
all_regions = north_region + south_region + east_region
print("Concatenated Temperatures:", all_regions)

# Iteration through readings
print("\nTemperature Analysis:")
for temp in all_regions:
    print(f"- Recorded {temp}°C")

# Count and index operations
target_temp = 21.5
print(f"\nTemperature Statistics for {target_temp}°C:")
print("Occurrences:", all_regions.count(target_temp))
print("First recorded at index:", all_regions.index(target_temp))

#Membership check
user_query = 22.1
if user_query in all_regions:
    print(f"\n{user_query}°C was recorded in our data")
else:
    print(f"\n{user_query}°C was not found in our records")

# Aggregate functions
print("\nSystem-wide Statistics:")
print("Maximum Temperature:", max(all_regions))
print("Minimum Temperature:", min(all_regions))
print("Total Temperature Sum:", sum(all_regions))
# You can also use any() and all()
```

### Reversing a Tuple

```python
steps = ("start", "load data", "process", "validate", "save", "end")

# reversed() returns an iterator
reversed_steps = tuple(reversed(steps))
print("Reversed tuple:", reversed_steps)

reversed_steps_slice = steps[::-1]
print("Reversed tuple (using slicing):", reversed_steps_slice)
```

### Performance: Looping over Large Reversed Tuples

```python
import time
huge_tuple = tuple(range(200_000_000)) # 200 million items

# Using reversed() - lazy
start = time.time()
for i in reversed(huge_tuple):
    break
end = time.time()
print("Time with reversed():", round(end - start, 4), "seconds")

# Using slicing - eager
start = time.time()
for i in huge_tuple[::-1]:
    break
end = time.time()
print("Time with slicing:", round(end - start, 4), "seconds")
```

### Sorting the Tuple Elements

```python
values = (42, 5, 12, 99, 18)

# sorted() returns a list
sorted_values = tuple(sorted(values))
print("Sorted values:", sorted_values)
```

### Sorting Tuples with the `key` Argument

```python
records = (
    ("sensor1", 30),
    ("sensor2", 25),
    ("sensor3", 40),
)

sorted_by_value = tuple(sorted(records, key=lambda x: x[1]))
print("Sorted by sensor reading:", sorted_by_value)
```

```python
products = (
    ("product_a", 120, 10),
    ("product_b", 200, 100),
    ("product_v", 150, 20),
) # (name, price, discount)

# Sort by final price after discount
sorted_by_final_price = tuple(sorted(products, key=lambda x: x[1] - x[2]))
print("Sorted by final price:", sorted_by_final_price)
```

### Tuple Aliases and Shallow Copies

Shallow copies and aliases of tuples that contain only immutable elements are semantically the same

```python
# List of enabled modules
modules = ("core", "auth", "storage")

# Alias points to the same tuple in memory
alias = modules
# Create a shallow copy using slicing
copied = modules[:]
# Or you can use the tuple constructor
# copied = tuple(modules)

print("modules is alias:", modules is alias)
print("modules is copied:", modules is copied)
```

### Tuple Deep Copies

```python
import copy

# Tuple of settings, each with a sub-list of values
settings = (["volume", 70], ["brightness", 50])

# Create a deep copy
deep_copy = copy.deepcopy(settings)

# Modify the inner list in the deep copy
deep_copy[0][1] = 20

print("Original after deep copy:", settings)
print("Deep copy:", deep_copy)
```

### Tuple Unpacking

```python
def compute_volume(length, width, height):
    return length * width * height

dimensions = (10, 5, 2)
length, width, height = dimensions

volume = compute_volume(length, width, height)
print("Volume:", volume)
```

### Tuple Unpacking in Function Calls

```python
# Unpacking with *
volume = compute_volume(*dimensions) # Same as: compute_volume(dimensions[0], dimensions[1], dimensions[2])
print("Volume:", volume)
```

### Packing Function Arguments into a Tuple

```python
# Packing with *
def sum_numbers(*args):
    print("args:", args)
    return sum(args)

print("Sum:", sum_numbers(1, 2, 3))
print("Sum:", sum_numbers(1, 2, 3, 4, 5))
```

### Combining Positional and Variable-length Arguments

```python
def generate_report(title, *sections):
    print(f"=== {title} ===")
    for i, section in enumerate(sections, 1):
        print(f"{i}. {section}")

generate_report("System Health Report",
               "CPU usage: 47%",
               "RAM usage: 68%",
               "Disk I/O: Normal")
```

### Returning Multiple Values from a Function

```python
def min_max(numbers):
    return min(numbers), max(numbers) # returns one value which is a tuple (in python a function can only return one single object)
    # return (min(numbers), max(numbers))
    
n = [1, 20, 33, 401, 5]

min_result, max_result = min_max(n)
print(f"Min is {min_result} and max is {max_result}")
```

### Extended Unpacking with `*`

```python
http_response = (200, "OK", "Data loaded", "X-Request-Id: abc123", "X-Latency: 42ms")

# The * variable captures everything in the middle
status_code, status_msg, body, *headers = http_response

print("Status:", status_code, status_msg)
print("Body:  ", body)
print("Headers:", headers)

# Grab just the first and last, discard the middle
first, *_, last = http_response
print("\nFirst:", first)
print("Last: ", last)
```

### Tuples as Dictionary Keys and Set Members

Tuples are hashable (when all their elements are immutable), so they can be used as dictionary keys or stored in sets — something lists cannot do.

```python
# (latitude, longitude) pairs as dict keys — lists would raise TypeError
distance_cache = {}
distance_cache[(40.7128, -74.0060), (34.0522, -118.2437)] = 3940  # km approx

print("Cached distance:", distance_cache[(40.7128, -74.0060), (34.0522, -118.2437)], "km")

# Tuples in a set — useful for deduplicating coordinate pairs
visited = set()
visited.add((40.7128, -74.0060))
visited.add((34.0522, -118.2437))
visited.add((40.7128, -74.0060))  # duplicate — silently ignored

print("Unique locations visited:", len(visited))

# A list cannot be a dict key or set member
try:
    bad = {[1, 2]: "value"}
except TypeError as e:
    print("List as key error:", e)
```

### Named Tuples

`collections.namedtuple` creates a tuple subclass whose fields are accessible by name, not just index — making code self-documenting without the overhead of a full class.

```python
from collections import namedtuple

# Define a Point type with named fields
Point = namedtuple("Point", ["x", "y"])

origin = Point(0, 0)
click  = Point(x=120, y=45)

# Access by name (readable) or by index (still works)
print("Click position:", click.x, click.y)
print("Same via index:", click[0], click[1])

# Still a tuple — immutable, hashable, unpackable
x, y = click
print("Unpacked:", x, y)
print("Is tuple:", isinstance(click, tuple))

# _replace() returns a new namedtuple with some fields changed
moved = click._replace(x=200)
print("Moved:", moved)
```
