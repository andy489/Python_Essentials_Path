## Fundamentals of File I/O in Python

### Reading and Writing TExt Files

```python
import os
from datetime import date,datetime

# Globomantics – Chat Transcript Logger

chat_transcript = [
    "Client: Hello, I need to check the status of my shipment.",
    "Bot: Sure, can I have your tracking number?",
    "Client: It's GLOBO123456",
    "Bot: Thank you. Your package is currently in transit and should arrive by Thursday.",
    "Client: Great, thanks!",
    "Bot: You're welcome. Have a nice day!"
]

TRANSCRIPTS_DIR = "transcripts"
file_path = os.path.join(TRANSCRIPTS_DIR, f"{date.today()}_chat_log.txt")

os.makedirs(TRANSCRIPTS_DIR, exist_ok=True)

with open(file_path, 'w', encoding='utf-8') as file:
    file.write("\n".join(chat_transcript) + "\n")

print(f"Chat transcript saved to {file_path}")

with open(file_path, 'r', encoding='utf-8') as file:
    contents = file.read()

print("\nTranscript Preview:\n")
print(contents)
```

### Safer File Access with Context Managers

```python
# Globomantics – Daily Metrics Logger

LOGS_DIR = "logs"
LOG_FILE = os.path.join(LOGS_DIR, "shipment_metrics.log")

daily_metrics = {
    "date": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
    "region": "West Coast Hub",
    "shipments_processed": 428,
    "delivery_failures": 7,
    "avg_transit_time_hrs": 26.4,
}

def format_log_entry(metrics: dict) -> str:
    return (
        f"[{metrics['date']}] "
        f"Region: {metrics['region']} | "
        f"Shipments: {metrics['shipments_processed']} | "
        f"Failures: {metrics['delivery_failures']} | "
        f"Avg Transit: {metrics['avg_transit_time_hrs']} hrs\n"
    )

def log_metrics(metrics: dict) -> None:
    os.makedirs(LOGS_DIR, exist_ok=True)
    entry = format_log_entry(metrics)
    with open(LOG_FILE, 'a', encoding='utf-8') as f:
        f.write(entry)
    print(f"Logged: {entry}", end="")

log_metrics(daily_metrics)
```

### Efficient Line-by-line File Reading

```python
# Globomantics – Access Log Processor (Line-by-line)

LOGS_DIR = "logs"
LOG_FILE = os.path.join(LOGS_DIR, "shipment_access.log")
ENTRY_COUNT = 100
PREVIEW_LINES = 5

def generate_log(path: str, count: int) -> None:
    os.makedirs(LOGS_DIR, exist_ok=True)
    with open(path, 'w', encoding='utf-8') as f:
        for i in range(1, count + 1):
            status = "FAILURE" if i % 10 == 0 else "SUCCESS"
            f.write(f"EventID={1000+i} | Status={status} | Hub=West | Timestamp=2025-07-24T10:{i%60:02d}:00Z\n")
    print(f"Simulated access log written to {path}")

def print_failures(path: str) -> None:
    print("\n=== Failed Shipments Detected ===")
    failed_count = 0
    with open(path, 'r', encoding='utf-8') as f:
        for line in f:
            if "FAILURE" in line:
                print(line.strip())
                failed_count += 1
    print(f"\nTotal failed shipment events: {failed_count}")

def print_preview(path: str, n: int = PREVIEW_LINES) -> None:
    print(f"\n=== First {n} Log Entries ===")
    with open(path, 'r', encoding='utf-8') as f:
        for _, line in zip(range(n), f):
            print(line.strip())

def print_total(path: str) -> None:
    with open(path, 'r', encoding='utf-8') as f:
        total = sum(1 for _ in f)
    print(f"\nTotal lines: {total}")

generate_log(LOG_FILE, ENTRY_COUNT)
print_failures(LOG_FILE)
print_preview(LOG_FILE)
print_total(LOG_FILE)
```
