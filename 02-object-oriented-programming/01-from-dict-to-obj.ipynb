## From Dictionaries to Objects

### Lists
Python lists are not restricted to one data type. A list stores references to Python objects, and those objects can be of different types.

```python
employee1 = ["Ji-Soo", 38, "developer", 1200]
employee2 = ["Lauren", 44, "tester", 1000]

employees = [employee1, employee2]
```

```python
type(employee1)
```

```python
print(type(employee1[0]))
print(type(employee1[1]))
```

```python
for e in employees:
    print(f"{e[0]}'s salary is ${e[3]:,.2f}")
```

## Maps

```python
employee1 = { 
    "name": "Ji-Soo", 
    "age": 38, 
    "position": "developer",
    "salary": 1200
}
employee2 = {
    "name": "Lauren",
    "age": 44,
    "position": "tester", 
    "salary": 1000
}

employees = [employee1, employee2]

for e in employees:
    print(f"{e['name']}'s salary is ${e['salary']:,.2f}")
```

```python
def increase_salary(employee, percent):
    employee['salary'] += employee['salary'] * (percent/100) 
```

```python
def employee_info(employee):
    print(
        f"{employee['name']} is {employee['age']} years old. "
        f"Employee is a {employee['position']} with the salary of ${employee['salary']:,.2f}"
    )
```

```python
increase_salary(employee1, 10)
employee_info(employee1)
```

```python
def init_employee(name, age, position, salary):
    return {
        "name": name,
        "age": age,
        "position": position,
        "salary": salary
    }
```

```python
employee3 = init_employee("Mateo", 38, "developer", 200)
```

```python
employee_info(employee3)
```

## Designing a Class

## Class vs Object

| Class | Object |
|---|---|
| A **blueprint/template** | An **instance** created from a class |
| Defines attributes and methods | Contains actual values for those attributes |
| Example: `Employee` | Example: `employee1 = Employee(...)` |
| Describes what an object should have/do | Represents a specific thing |

**Simple analogy:**  
**Class = blueprint of a house**  
**Object = actual house built from the blueprint**

```python
s = "hello"
i = 2
l = [1, 2, True]
d = {"a": 3}
```

```python
print(s.__class__)
print(i.__class__)
print(l.__class__)
print(d.__class__)
```
