## Underscore Rules

| Convention | Meaning | Example |
|---|---|---|
| `_name` | Internal/private by convention | `_salary` |
| `__name` | Name mangling | `__salary` |
| `__name__` | Special Python method/attribute (dunder) | `__init__`, `__str__` |

- `_name` → "Please treat this as internal."
- `__name` → Python modifies the name to reduce accidental access/overriding.
- `__name__` → Reserved for Python's special methods and attributes.

### Dunder Methods

**Dunder** = **double underscore** (`__`).

They are special Python methods with **two underscores before and after the name**.

```python
__init__
__str__
__len__
__eq__
```
Dunder methods are Python's way of defining special behavior for objects. So you can think of dunder methods as **hooks that tell Python how your objects should behave with built-in operations.**

````python
x = 2
y = 2
l = [1, 2]

# Dunder methods

## Adding two objects
x + y     # Internally calls: x.__add__(y)

# Converting to string for printing
print(x)  # Internally calls: str(x), which calls x.__str__()

# Getting the length of a container
len(l)    # Internally calls: x.__len__()

# Comparing two objects
x == y    # Internally calls: x.__eq__(y)
````

````python
class Employee:
    def __init__(self, name, age, position, salary):
        self.name = name
        self.age = age
        self.position = position
        self.salary = salary

    def increase_salary(self, percent):
        self.salary += self.salary * (percent/100) 

    def __str__(self):
        return (
            f"{self.name} is {self.age} years old. "
            f"Employee is a {self.position} with the salary of ${self.salary:,.2f}"
        )

e3 = Employee("Pesho", 33, "dev-ops", 1500)
````

````python
from dataclasses import dataclass
````

````python
@dataclass
class EmployeeDataClass:
    name: str
    age: int
    position: str
    salary: float

    def increase_salary(self, percent):
        self.salary += self.salary * (percent/100)

e3 = EmployeeDataClass("Pesho", 33, "dev-ops", 1500)
print(e3)
````
